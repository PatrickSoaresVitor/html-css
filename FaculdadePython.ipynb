### Escrever

#### No python usa print para escrever e input para ler

```python
print('Hello Word')

nome = input('Qual é o seu nome?\n')
print('Olá %s!' %nome)
print('Olá {0}!' .format(nome))
print(f'Olá {nome}!')
print('Olá', nome,'!') # Não usar
```

Existe varias formas de escrever em python, podendo usar o com %|s,d,f|, com o .format e o f'{nome}, para poder formatar basta colocar : e formatar depois dos dois pontos. ! E NÃO USAR A FORMA PADRÃO !. O input retorna de padrão uma string então colocamos int ou float antes, sendo int para numero inteiro o float para numeros com casas decimais

```python
nome = input('Nome: ')
idade = int(input('Idade: '))
nota = float(input('Digite a nota:'))

print('%s possui %d anos e nota %.2f' % (nome, idade, nota))
print(nome, 'possui', idade, 'anos e nota', nota) # Não usar
print('{0} possui {1} anos e nota {2:.2f}' .format(nome, idade, nota))
print('{} possui {} anos e nota {:.2f}' .format(nome, idade, nota))
print(f'{nome} possui {idade} anos e nota {nota:.2f}')
```

#### 5- Fazer um algoritmo que calcule a média de um aluno na disciplina de ALP. Para isso solicite o nome do aluno, a nota da prova A e a nota da prova B. Sabe-se que a nota da prova A tem peso 2 e a nota da prova B peso 1. Mostre a média e o nome do aluno como resultado.

```python
nome = input('Digite o nome do aluno: ')
notaA = int(input('Digite a nota da prova A: '))
notaB = int(input('Digite a nota da prova B: '))
media = (notaA * 2 + notaB * 1) / 3
print('A média do {0} é de {1:.2f}' .format(nome, media))
```

####6- Construa um algoritmo que, tendo como entrada dois pontos quaisquer do plano, imprima a distância entre eles

```python
import math
ponto1x = int(input('Digite o x do primeiro ponto: '))
ponto1y = int(input('Digite o y do primeiro ponto: '))
ponto2x = int(input('Digite o x do segundo ponto: '))
ponto2y = int(input('Digite o y do segundo ponto: '))
distanciaP = math.sqrt((ponto2x - ponto1x)**2 + (ponto2y - ponto1y)**2)
print('A distancia entre os dois pontos é de {0}' .format(distanciaP))
```

####7- Prepare um algoritmo capaz de inverter um número, de 3 dígitos, fornecido, ou seja, apresentar primeiro a unidade e, depois, a dezena e a centena.

```python
numero = int(input('Digite um numero de três digitos: '))
unidade = numero // 100
dezena = numero % 100 //10
centena =  numero %10
numeroinv = centena * 100 + dezena * 10 + unidade
print('o {0} invertido é {1}' .format(numero, numeroinv))
```

####4- Fazer um algoritmo que leia 3 valores inteiros e verifique se eles podem formar um triângulo. Se for possível formar um triângulo, escreva uma mensagem informando se é um triângulo equilátero, isósceles ou escaleno.
Observações:
  - O comprimento de um lado do triângulo é sempre menor do que a soma dos outros dois.
  - Equilátero: todos lados iguais
  - Isósceles: dois lados iguais
  - Escaleno: todos os lados diferentes

```python
lado1 = int(input('Digite um lado do triângulo: '))
lado2 = int(input('Digite um lado do triângulo: '))
lado3 = int(input('Digite um lado do triângulo: '))
if lado1 < (lado2 +lado3) and lado2 < (lado1 +lado3) and lado3 < (lado2 +lado1):
  if lado1 == lado2 and lado2 == lado3:
    print('É um triangulo Equilátero')
  elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print('É um triangulo Isóceles')
  else:
    print('É um triangulo Escaleno')
else:
  print('Não é um triângulo')
```
